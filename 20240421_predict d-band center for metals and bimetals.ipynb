```python
for i in df.index:
    for j in df.columns:
        if(i =='Fe'and j =='Fe'):
            print ('host '+i+', guest '+j+', val' +str(df.loc[i,j]))
```

```python
for i in df.index:
    for j in df.columns:
        if(i != j):
           df.loc[i,j] += df.loc[i,i]
df
```

```python
with open ('data_impurities_features.csv', encoding = 'UTF-8') as f:
    feat = pd.read_csv(f, index_col = 0)
feat
```

```python
feat.loc ['Co']
```

```python
feat.head(2)
```

```python
feat = feat.drop ('name', axis = 'columns')
feat
```

```python
x = list ()
y = list ()
for i in df.index:
    for j in df.columns:
        vec_i = feat.loc[i].to_numpy()
        vec_j = feat.loc[j].to_numpy()
        x_val = np.concatenate ((vec_i, vec_j))
        y_val = df.loc [i,j]
        x.append (x_val)
        y.append (y_val)
        if i == 'Fe' and j == 'Co':
            print ('host '+ i +', guest '+ j +', input = ', str(x_val) +', output = ' + str(y_val))
```

```python
X = np.array(x)
y = np.array(y)
X.shape, y.shape
```

```python
from sklearn.utils import shuffle
a = np.array(range(10))
print (a)
a = shuffle (a)
print (a)
```

```python
X_r, y_r = shuffle(X,y)
X_train, y_train = X_r[:-30,:], y_r[:-30]
X_test, y_test = X_r[-30:,:], y_r[-30:]
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
y_pred_train_lr = lr.predict(X_train)
y_pred_test_lr = lr.predict (X_test) 
```

```python
import matplotlib.pyplot as plt
plt.figure (figsize = (6, 6))
plt.scatter (y_train, y_pred_train_lr, alpha = 0.5, color = 'b', label = 'training')
plt.scatter (y_test, y_pred_test_lr, alpha = 0.5, color = 'r', label = 'test')
plt.legend()
plt.xlabel ('DFT')
plt.ylabel ('prediction')
```

```python
from sklearn.metrics import mean_squared_error
rmse_tr_lr = mean_squared_error(y_train, y_pred_train_lr, squared = False)
rmse_te_lr = mean_squared_error (y_test, y_pred_test_lr, squared = False)
print ('RMSE(training)%.3f' %rmse_tr_lr)
print ('RMSE(test)%.3f' %rmse_te_lr)
```

```python
from sklearn.model_selection import cross_val_score
r2_scores_lr = cross_val_score (lr, X, y, scoring = 'r2', cv = 5)
rmse_scores_lr = cross_val_score (lr, X, y, scoring = 'neg_root_mean_squared_error', cv = 5)
print ('cv result:')
print ('Fold: %i, mean R2: %.3f' % (len (r2_scores_lr), r2_scores_lr.mean()))
print ('Fold: %i, mean RMSE: %.3f' % (len (rmse_scores_lr), -rmse_scores_lr.mean()))
```

```python
# use KFold for cross validation
from sklearn.model_selection import cross_val_score, KFold

crossvalidation = KFold(n_splits = 10, shuffle = True)
r2_scores_lr = cross_val_score (lr, X, y, scoring = 'r2', cv = crossvalidation)
rmse_scores_lr = cross_val_score (lr, X, y, scoring = 'neg_root_mean_squared_error', cv = crossvalidation)
print ('cv result:')
print ('Fold: %i, mean R2: %.3f' % (len (r2_scores_lr), r2_scores_lr.mean()))
print ('Fold: %i, mean RMSE: %.3f' % (len (rmse_scores_lr), -rmse_scores_lr.mean()))
```

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
# gradient boosting regression
gbr = GradientBoostingRegressor () 
gbr.fit(X_train, y_train)
y_pred_train_gbr = gbr.predict (X_train)
y_pred_test_gbr = gbr.predict (X_test)
```

```python
plt.figure(figsize = (5,5))
plt.scatter (y_train, y_pred_train_gbr, alpha = 0.5, color = 'b', label = 'training')
plt.scatter (y_test, y_pred_test_gbr, alpha = 0.5, color = 'r', label = 'test')
plt.legend()
plt.xlabel ('DFT')
plt.ylabel ('prediction')
```

```python
from sklearn.metrics import mean_squared_error
rmse_tr_gbr = mean_squared_error(y_train, y_pred_train_gbr, squared = False)
rmse_te_gbr = mean_squared_error (y_test, y_pred_test_gbr, squared = False)
print ('RMSE(training)%.3f' %rmse_tr_gbr)
print ('RMSE(test)%.3f' %rmse_te_gbr)
```

```python
#cross validation
from sklearn.model_selection import cross_val_score, KFold

crossvalidation = KFold(n_splits = 10, shuffle = True)
r2_scores_gbr = cross_val_score (gbr, X, y, scoring = 'r2', cv = crossvalidation)
rmse_scores_gbr = cross_val_score (gbr, X, y, scoring = 'neg_root_mean_squared_error', cv = crossvalidation)
print ('cv result:')
print ('Fold: %i, mean R2: %.3f' % (len (r2_scores_gbr), r2_scores_gbr.mean()))
print ('Fold: %i, mean RMSE: %.3f' % (len (rmse_scores_gbr), -rmse_scores_gbr.mean()))
```
